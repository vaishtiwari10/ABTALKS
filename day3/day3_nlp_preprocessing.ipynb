```python
# tokenize a paragraph using nltk and spacy
from nltk.tokenize import word_tokenize
import spacy

text = "Natural Language Processing enables computers to understand human language. It's used in search engines, chatbots, and spam filters."

print("Original text:")
print(text)

# nltk tokenization
tokens_nltk = word_tokenize(text)
print(f"\nNLTK tokens ({len(tokens_nltk)}):")
print(tokens_nltk)

# spacy tokenization
nlp = spacy.load("en_core_web_sm")
doc = nlp(text)
tokens_spacy = [token.text for token in doc]
print(f"\nspaCy tokens ({len(tokens_spacy)}):")
print(tokens_spacy)
```

```python
# remove stop words and punctuation
from nltk.corpus import stopwords
import string

stop_words = set(stopwords.words('english'))

cleaned = [t.lower() for t in tokens_nltk if t.lower() not in stop_words and t not in string.punctuation]

print(f"Before ({len(tokens_nltk)} tokens): {tokens_nltk}")
print(f"After  ({len(cleaned)} tokens): {cleaned}")
```

```python
# lemmatize
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
lemmatized = [lemmatizer.lemmatize(w) for w in cleaned]

print(f"Before: {cleaned}")
print(f"After:  {lemmatized}")
```

```python
# bag of words
from sklearn.feature_extraction.text import CountVectorizer

corpus = [
    "The quick brown fox jumps over the lazy dog",
    "Natural language processing enables computers to understand human language",
    "I loved this product it is the best purchase I have made",
    "Scientists discover new species of deep sea fish off the coast of Japan",
    "The meeting is scheduled for Monday please bring your laptops"
]

vectorizer = CountVectorizer(stop_words='english')
bow = vectorizer.fit_transform(corpus)

print(f"Vocabulary: {vectorizer.get_feature_names_out()}")
print(f"\nBoW matrix ({bow.shape[0]} docs x {bow.shape[1]} features):")
print(bow.toarray())
```
