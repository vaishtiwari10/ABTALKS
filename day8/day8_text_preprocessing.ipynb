# Text Preprocessing Pipeline

```python
import re
import nltk
import pandas as pd

for r in ['punkt', 'punkt_tab', 'stopwords', 'wordnet', 'omw-1.4']:
    nltk.download(r, quiet=True)

from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()
stop_words = set(stopwords.words('english'))
```

## loading raw text samples

```python
social_media = (
    "OMG!! Just tried the NEW cafe on 5th Ave and it's ABSOLUTELY amazing!!! "
    "The iced latte was sooo good. Can't believe I haven't been here before... "
    "#CoffeeLover #BestCafeEver @coffeehouse_nyc Check it out ASAP!!"
)

news_article = (
    "The Federal Reserve announced on Wednesday that it would be raising interest rates "
    "by 0.25 percentage points, marking the 10th consecutive increase since March 2022. "
    "\"We are committed to bringing inflation down to our 2% target,\" said Chair Jerome Powell."
)

customer_review = (
    "I bought this laptop 3 weeks ago and I'm EXTREMELY disappointed!!! The battery life "
    "is terrible - it barely lasts 2 hours?! The keyboard feels cheap and the trackpad "
    "is unresponsive... Worst purchase I've EVER made. DO NOT buy this!!"
)

samples = {
    'Social Media': social_media,
    'News Article': news_article,
    'Customer Review': customer_review
}

for name, text in samples.items():
    print(f'--- {name} ---')
    print(text, '\n')
```

## step 1 - tokenisation

```python
for name, text in samples.items():
    print(f'--- {name} ---')
    sentences = sent_tokenize(text)
    words = word_tokenize(text)
    print(f'Sentences ({len(sentences)}):')
    for i, s in enumerate(sentences, 1):
        print(f'  {i}. {s}')
    print(f'Word tokens ({len(words)}): {words}\n')
```

## step 2 - stopword removal

```python
for name, text in samples.items():
    print(f'--- {name} ---')
    words = [w.lower() for w in word_tokenize(text)]
    filtered = [w for w in words if w not in stop_words]
    print(f'Before: {len(words)} tokens, {len(set(words))} unique')
    print(f'After:  {len(filtered)} tokens, {len(set(filtered))} unique')
    print(f'Removed: {sorted(set(words) - set(filtered))}\n')
```

## step 3 - punctuation stripping

```python
# keeps only letters, numbers, spaces
# needed because punctuation creates fake vocab entries
punct_pattern = re.compile(r'[^a-zA-Z0-9\s]')

for name, text in samples.items():
    print(f'--- {name} ---')
    before = word_tokenize(text)
    after = word_tokenize(punct_pattern.sub('', text))
    print(f'Before: {len(before)} tokens -> After: {len(after)} tokens')
    print(f'Clean: {after}\n')
```

## step 4 and 5 - stemming vs lemmatisation

```python
for name, text in samples.items():
    print(f'--- {name} ---')
    clean = punct_pattern.sub('', text.lower())
    tokens = [w for w in word_tokenize(clean) if w not in stop_words and len(w) > 1]

    stemmed = [stemmer.stem(w) for w in tokens]
    lemmatized = [lemmatizer.lemmatize(w) for w in tokens]

    df = pd.DataFrame({'original': tokens, 'stemmed': stemmed, 'lemmatized': lemmatized})
    print(df.to_string(), '\n')
```

## full pipeline comparison

```python
for name, text in samples.items():
    print(f'===== {name} =====')
    words = word_tokenize(text)
    print(f'1. Tokenised       -> {len(sent_tokenize(text))} sentences, {len(words)} words')

    lower = [w.lower() for w in words]
    no_stop = [w for w in lower if w not in stop_words]
    print(f'2. Stopwords gone  -> {len(lower)} -> {len(no_stop)} tokens')

    no_punct = [w for w in [punct_pattern.sub('', w) for w in no_stop] if w]
    print(f'3. Punctuation out -> {len(no_stop)} -> {len(no_punct)} tokens')

    stems = [stemmer.stem(w) for w in no_punct]
    lemmas = [lemmatizer.lemmatize(w) for w in no_punct]
    print(f'4. Stemmed         -> {stems}')
    print(f'5. Lemmatised      -> {lemmas}\n')
```

## why each step matters

| step | what it does | what breaks without it |
|------|-------------|------------------------|
| tokenisation | splits text into words and sentences | model gets one big string, can't extract features |
| stopword removal | removes filler words like the, is, and | common words dominate counts, hide important terms |
| punctuation stripping | removes symbols and special chars | good and good!!! treated as different words |
| stemming | chops endings to get root form | running, runs, ran seen as 3 separate words |
| lemmatisation | dictionary lookup for base word | same issue but lemmatisation gives real words unlike stemming |
